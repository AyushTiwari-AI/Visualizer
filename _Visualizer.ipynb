```python
import pandas as pd
import matplotlib.pyplot as plt




import matplotlib.pyplot as plt


df = None
current_fig = None


def load_dataset():
    global df
    print("\n== Load Dataset ==")
    path = input("Enter the path of the dataset (CSV file): ")
    try:
        df = pd.read_csv(path)
        print("Dataset loaded successfully!")
    except Exception as e:
        print(f"Error loading dataset: {e}")


def explore_data():
    global df
    if df is None:
        print("\nNo dataset loaded! Please load a dataset first.")
        return

    print("\n== Explore Data ==")
    print("1. Display the first 5 rows")
    print("2. Display the last 5 rows")
    print("3. Display column names")
    print("4. Display data types")
    print("5. Display basic info")
    choice = input("Enter your choice: ")

    if choice == "1":
        print("\n", df.head())
    elif choice == "2":
        print("\n", df.tail())
    elif choice == "3":
        print("\nColumn names:", list(df.columns))
    elif choice == "4":
        print("\n", df.dtypes)
    elif choice == "5":
        print()
        df.info()
    else:
        print("Invalid choice!")


def dataframe_operations():
    global df
    if df is None:
        print("\nNo dataset loaded! Please load a dataset first.")
        return

    print("\n== Perform DataFrame Operations ==")
    print("1. Sort by column")
    print("2. Filter rows by condition")
    print("3. Select specific columns")
    print("4. Group by column and aggregate")
    print("5. Rename columns")
    print("6. Drop a column")
    choice = input("Enter your choice: ")

    try:
        if choice == "1":
            col = input("Enter column name to sort by: ")
            order = input("Ascending order? (y/n): ").strip().lower()
            df = df.sort_values(by=col, ascending=(order != "n"))
            print(f"\nDataFrame sorted by '{col}'.")
            print(df.head())
        elif choice == "2":
            col = input("Enter column name for condition: ")
            op = input("Enter operator (==, !=, >, <, >=, <=): ")
            val = input("Enter value to compare: ")
            try:
                val_cast = float(val)
            except ValueError:
                val_cast = val
            query_str = f"`{col}` {op} @val_cast"
            result = df.query(query_str)
            print(f"\nFiltered rows where {col} {op} {val}:")
            print(result)
        elif choice == "3":
            cols = input("Enter column names separated by commas: ")
            cols_list = [c.strip() for c in cols.split(",")]
            print(f"\nSelected columns:\n{df[cols_list]}")
        elif choice == "4":
            group_col = input("Enter column name to group by: ")
            agg_col = input("Enter column name to aggregate: ")
            agg_func = input("Enter aggregation function (sum, mean, count, max, min): ")
            result = df.groupby(group_col)[agg_col].agg(agg_func)
            print(f"\nGrouped by '{group_col}', aggregated '{agg_col}' with '{agg_func}':")
            print(result)
        elif choice == "5":
            old_name = input("Enter current column name: ")
            new_name = input("Enter new column name: ")
            df = df.rename(columns={old_name: new_name})
            print(f"\nColumn '{old_name}' renamed to '{new_name}'.")
            print(df.head())
        elif choice == "6":
            col = input("Enter column name to drop: ")
            df = df.drop(columns=[col])
            print(f"\nColumn '{col}' dropped.")
            print(df.head())
        else:
            print("Invalid choice!")
    except Exception as e:
        print(f"Error performing operation: {e}")


def handle_missing_data():
    global df
    if df is None:
        print("\nNo dataset loaded! Please load a dataset first.")
        return

    print("\n== Handle Missing Data ==")
    print("1. Display rows with missing values")
    print("2. Fill missing values with mean")
    print("3. Drop rows with missing values")
    print("4. Replace missing values with a specific value")
    choice = input("Enter your choice: ")

    if choice == "1":
        missing_rows = df[df.isnull().any(axis=1)]
        if missing_rows.empty:
            print("\nNo missing values found in the dataset!")
        else:
            print("\nRows with missing values:")
            print(missing_rows)
    elif choice == "2":
        numeric_cols = df.select_dtypes(include="number").columns
        df[numeric_cols] = df[numeric_cols].fillna(df[numeric_cols].mean())
        print("\nMissing numeric values filled with column mean.")
    elif choice == "3":
        before = len(df)
        df = df.dropna()
        after = len(df)
        print(f"\nDropped {before - after} rows with missing values.")
    elif choice == "4":
        val = input("Enter the value to replace missing data with: ")
        df = df.fillna(val)
        print(f"\nMissing values replaced with '{val}'.")
    else:
        print("Invalid choice!")


def descriptive_statistics():
    global df
    if df is None:
        print("\nNo dataset loaded! Please load a dataset first.")
        return

    print("\n== Generate Descriptive Statistics ==")
    print(df.describe(include="all"))


def data_visualization():
    global df, current_fig
    if df is None:
        print("\nNo dataset loaded! Please load a dataset first.")
        return

    print("\n== Data Visualization ==")
    print("1. Bar Plot")
    print("2. Line Plot")
    print("3. Scatter Plot")
    print("4. Pie Chart")
    print("5. Histogram")
    print("6. Stack Plot")
    choice = input("Enter your choice: ")

    try:
        fig, ax = plt.subplots(figsize=(8, 5))

        if choice == "1":
            x_col = input("Enter x-axis column name: ")
            y_col = input("Enter y-axis column name: ")
            print("\nGenerating bar plot...")
            ax.bar(df[x_col], df[y_col])
            ax.set_xlabel(x_col)
            ax.set_ylabel(y_col)
            ax.set_title(f"Bar Plot: {y_col} vs {x_col}")
            print("Bar plot displayed successfully!")

        elif choice == "2":
            x_col = input("Enter x-axis column name: ")
            y_col = input("Enter y-axis column name: ")
            print("\nGenerating line plot...")
            ax.plot(df[x_col], df[y_col], marker="o")
            ax.set_xlabel(x_col)
            ax.set_ylabel(y_col)
            ax.set_title(f"Line Plot: {y_col} vs {x_col}")
            print("Line plot displayed successfully!")

        elif choice == "3":
            x_col = input("Enter x-axis column name: ")
            y_col = input("Enter y-axis column name: ")
            print("\nGenerating scatter plot...")
            ax.scatter(df[x_col], df[y_col])
            ax.set_xlabel(x_col)
            ax.set_ylabel(y_col)
            ax.set_title(f"Scatter Plot: {y_col} vs {x_col}")
            print("Scatter plot displayed successfully!")

        elif choice == "4":
            label_col = input("Enter column name for labels: ")
            value_col = input("Enter column name for values: ")
            print("\nGenerating pie chart...")
            ax.pie(df[value_col], labels=df[label_col], autopct="%1.1f%%")
            ax.set_title(f"Pie Chart: {value_col} by {label_col}")
            print("Pie chart displayed successfully!")

        elif choice == "5":
            col = input("Enter column name for histogram: ")
            bins = input("Enter number of bins (default 10): ")
            bins = int(bins) if bins.strip() else 10
            print("\nGenerating histogram...")
            ax.hist(df[col], bins=bins, edgecolor="black")
            ax.set_xlabel(col)
            ax.set_ylabel("Frequency")
            ax.set_title(f"Histogram of {col}")
            print("Histogram displayed successfully!")

        elif choice == "6":
            x_col = input("Enter x-axis column name: ")
            y_cols = input("Enter y-axis column names separated by commas: ")
            y_cols_list = [c.strip() for c in y_cols.split(",")]
            print("\nGenerating stack plot...")
            ax.stackplot(df[x_col], *[df[c] for c in y_cols_list], labels=y_cols_list)
            ax.set_xlabel(x_col)
            ax.legend(loc="upper left")
            ax.set_title("Stack Plot")
            print("Stack plot displayed successfully!")

        else:
            print("Invalid choice!")
            plt.close(fig)
            return

        current_fig = fig
        plt.show()

    except Exception as e:
        print(f"Error generating visualization: {e}")


def save_visualization():
    global current_fig
    print("\n== Save Visualization ==")
    if current_fig is None:
        print("No visualization to save! Please generate a plot first.")
        return
    filename = input("Enter file name to save the plot (e.g., scatter_plot.png): ")
    try:
        current_fig.savefig(filename)
        print(f"Visualization saved as {filename} successfully!")
    except Exception as e:
        print(f"Error saving visualization: {e}")


def main_menu():
    while True:
        print("\n========== Data Analysis & Visualization Program ==========")
        print("Please select an option:")
        print("1. Load Dataset")
        print("2. Explore Data")
        print("3. Perform DataFrame Operations")
        print("4. Handle Missing Data")
        print("5. Generate Descriptive Statistics")
        print("6. Data Visualization")
        print("7. Save Visualization")
        print("8. Exit")
        print("")

        choice = input("Enter your choice: ")

        if choice == "1":
            load_dataset()
        elif choice == "2":
            explore_data()
        elif choice == "3":
            dataframe_operations()
        elif choice == "4":
            handle_missing_data()
        elif choice == "5":
            descriptive_statistics()
        elif choice == "6":
            data_visualization()
        elif choice == "7":
            save_visualization()
        elif choice == "8":
            print("\nExiting the program. Goodbye!")
            break
        else:
            print("\nInvalid choice! Please select a valid option (1-8).")



main_menu()
```
